```python
from openai import OpenAI

client = OpenAI(base_url="http://128.118.54.16:11434/v1", api_key="no-api")

reply = ""
stream = client.chat.completions.create(
    model="gpt-oss:120b",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Write Python code to compute a convex hull of a set of points. Be concise and do not add any explanation"}
    ],
    stream=True
)

for chunk in stream:
    if chunk.choices[0].delta.content:
        content = chunk.choices[0].delta.content
        print(content, end="", flush=True)
        reply += content

print(f"\n\nFinal output:\n{reply}")
```

```python
SYSTEM_PROMPT = """You are experience in writing an abstract for scientific and academic paper.
From rough writing and unarticulated texts, write a concise and logical abstract of about 250 words following this structure:
- Background: Provide a brief context for the research and explain what is already known about the topic. Be brief and maybe have it in 2 sentences.
- Purpose/Problem: State the specific problem, question, or aim of the study and why it is important to address it.
- Method/Framework components: Briefly describe the basic design, approach, or methods used in the research.
- Results: Summarize the most important findings or significant data from the study.
- Conclusion: Briefly explain the interpretation of the results, their significance, and their implications.


Rules:
- Just help me polish by improving the grammatical structure.
- Be self-contained: The abstract should be understandable on its own without the reader needing to refer to the full paper.
- Use clear language: Avoid jargon where possible, or ensure technical terms are understood by a wider audience in the field.
- Structure is key: Organize the information logically.
- Please expand only neccessary.
"""

USER_TEMPLATE = """User rough texts:
{user_query}

Context hints:
- Keep tone academic.
- The topic is mainly on geographic modeling. Here is the topic : Autonomous Geographic Modeling: Towards Geographic Discovery
"""


user_text = """
Geographic modeling is a crucial process in understanding the where, how and why of  complex phenomena. Traditionally, it rely on human-driven process which involve hypothesis formulation, model selection and execution, spatial pattern analysis, and result interpretation is iterative in nature and can take lot of time. Which makes human to focus on many things and time consuming and thereby can slow down or limit the process of knowledge discover. Recent advancements in Large Language Models (LLMs) have demonstrated capabilities in reasoning, tool use, and workflow automation. This makes it potentially be a gamechanger for performing this rigorous process.
In this study, an autonomous geographic modeling framework was developed. This framework leveraging the capabilities of LLM (such as GPT), aims at facilitatating the process of geographic knowledge discovery. Aa prototype system was developed to demonstrate this framwork. This prototype system comprises of muti-agents (researchquestion understanding agent, datagent, modeling agent, and result interpretation agent, and reporting agent.) orchestrated to work collectively to achieve all geographic modeling steps. We applied our prototype to three case studies. For all the cases, our prototype performed effectively. Although, still at its early stage, this study demonstrates a potential direction towards achieving autonomous geographic modeling thereby advancing toward AI-assisted scientific exploration and autonomous geographic discovery.


"""
```

```python
from openai import OpenAI

client = OpenAI(base_url="http://128.118.54.16:11434/v1", api_key="no-api")

reply = ""
stream = client.chat.completions.create(
    model="gpt-oss:120b",
    messages=[
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": USER_TEMPLATE.format(user_query=user_text)}
    ],
    stream=True
)

for chunk in stream:
    if chunk.choices[0].delta.content:
        content = chunk.choices[0].delta.content
        print(content, end="", flush=True)
        reply += content

print(f"\n\nFinal output:\n{reply}")
```
