```python
# Get Feedbacks GIS Copilot

import requests
import json
import pandas as pd
from datetime import datetime

def get_feedback(service, developer_api_key):
    url = f"https://www.gibd.online/api/download_feedback"

    data = {"service":service,
            "dev_api_key": developer_api_key}

    # Send POST request
    response = requests.post(
        url,
        headers={"Content-Type": "application/json"},
        json=data
    )
    # Handle response
    if response.status_code == 200:
        result = response.json()
        print("Data retrieved successfully!")
        print(f"Total records: {result['count']}")
        # print(json.dumps(result, indent=2))
    else:
        print(f"Error {response.status_code}: {response.text}")

    return result



def save_feedbacks(service, developer_api_key):
    result = get_feedback(service, developer_api_key)
    # Option 1: Save to file instead of printing
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    service_name = service.replace(" ", "_")
    filename = f"{service_name}_feedback_{timestamp}.json"

    with open(filename, 'w') as f:
        json.dump(result, f, indent=2)
    print(f"Data saved to: {filename}")


def show_feedbacks_summary(service, developer_api_key):
    result = get_feedback(service, developer_api_key)
    # Option 2: Show summary statistics instead of raw data
    if 'feedback' in result and isinstance(result['feedback'], list):
        feedback_data = result['feedback']
        print(f"\nSummary:")
        print(f"- Total feedback entries: {len(feedback_data)}")

        # Show first few entries as preview
        print(f"\nFirst 3 entries (preview):")
        for i, entry in enumerate(feedback_data[:3]):
            print(f"Entry {i+1}: {json.dumps(entry, indent=2)}")

    return

def feedbacks_csv(service, developer_api_key):
    result = get_feedback(service, developer_api_key)
    if 'feedback' in result and isinstance(result['feedback'], list) and len(result['feedback']) > 0:
        try:
            df = pd.DataFrame(result['feedback'])
            print(f"\nDataFrame Info:")
            print(f"Shape: {df.shape}")
            print(f"Columns: {list(df.columns)}")

            # Save as CSV for easier analysis
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            service_name = service.replace(" ", "_")
            csv_filename = f"{service_name}_feedback_{timestamp}.csv"
            df.to_csv(csv_filename, index=False)
            print(f"Data also saved as CSV: {csv_filename}")

            # # Show basic statistics
            # print(f"\nBasic Statistics:")
            # df.describe(include='all')

        except Exception as e:
            print(f"Could not convert to DataFrame: {e}")
    return df


def get_unique_values(info,service, developer_api_key):
    df = feedbacks_csv(service, developer_api_key)
    a = df[f'{info}'].unique()
    return a







```
