## GIS Copilot

```python
# Send Feedback

import requests
import json

sample_data = {
  "service_name": "GIS Copilot",
  "question_id": "q-xJGTxT5aOolF",
  "question": "What is the total length of rivers inside the national park boundary?",
  "feedback": "Good result",
  "feedback_message": "The user accepted the generated code.",
  "generated_code": "Import pandas\n ...",
  "data_overview": "User provided 'rivers.shp' (Polyline) and 'park.shp' (Polygon).",
  "task_breakdown": "task_breakdown",
  "selected_tools": "intersect",
  "workflow": "html code"
}


user_api_key = "gibd-services-ojXVx191ZNgVPdlmivsOtXsW"
url = f"https://www.gibd.online/api/feedback/{user_api_key}"


# Send POST request
response = requests.post(
    url,
    headers={"Content-Type": "application/json"},
    json=sample_data
)

# Handle response
if response.status_code == 200:
    result = response.json()
    print("Record created successfully!")
    print(json.dumps(result, indent=2))
else:
    print(f"Error {response.status_code}: {response.text}")
```
